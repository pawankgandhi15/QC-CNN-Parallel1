```python
# ============================================================
# 0.4 -- Clone Repository from GitHub
# ============================================================
import os
import subprocess
import sys

REPO_URL   = f"https://x-access-token:{GITHUB_TOKEN}@github.com/{GITHUB_USER}/{GITHUB_REPO}.git"
CLONE_DIR  = os.path.join(os.path.expanduser("~"), GITHUB_REPO)
IMPL_DIR   = os.path.join(CLONE_DIR, "implementation")

def run_cmd(cmd, cwd=None):
    """Run a shell command and print output."""
    result = subprocess.run(cmd, shell=True, cwd=cwd, text=True, capture_output=True)
    if result.stdout.strip(): print(result.stdout.strip())
    if result.stderr.strip(): print(result.stderr.strip())
    return result

if os.path.exists(CLONE_DIR):
    print(f"Repo already exists at {CLONE_DIR}. Pulling latest...")
    run_cmd(f"git remote set-url origin {REPO_URL}", cwd=CLONE_DIR)
    run_cmd("git pull origin main", cwd=CLONE_DIR)
else:
    print(f"Cloning {GITHUB_USER}/{GITHUB_REPO} ...")
    run_cmd(f"git clone {REPO_URL} {CLONE_DIR}")

# Set git identity
run_cmd(f'git config user.name "{GIT_NAME}"', cwd=CLONE_DIR)
run_cmd(f'git config user.email "{GIT_EMAIL}"', cwd=CLONE_DIR)

# Add to Python path and set working directory
if IMPL_DIR not in sys.path:
    sys.path.insert(0, IMPL_DIR)
os.chdir(IMPL_DIR)

print(f"Working directory: {os.getcwd()}")
print(f"Files: {[f for f in os.listdir('.') if not f.startswith('_')]}")
```

```python
# ============================================================
# 0.5 -- GitHub Push Helper
# ============================================================
import datetime

def push_results_to_github(message: str):
    """Stage all new/modified result files and push to GitHub."""
    if not AUTO_PUSH:
        print("AUTO_PUSH=False -- skipping push.")
        return
    print(f"\nPushing results to GitHub...")
    run_cmd("git add -A", cwd=CLONE_DIR)
    ts = datetime.datetime.utcnow().strftime("%Y-%m-%d %H:%M UTC")
    run_cmd(f'git commit -m "{message} [{ts}]" --allow-empty', cwd=CLONE_DIR)
    run_cmd("git push origin main", cwd=CLONE_DIR)
    print(f"Pushed: {message}")

print("Push helper ready. Will auto-push results after each experiment.")
```

```python
# ============================================================
# Experiment 1 -- Display Results
# ============================================================
import json
from pathlib import Path
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['figure.figsize'] = (12, 5)

# Table 2: Circuit metrics
t2_path = Path("results/experiment1/table2_circuit_metrics.json")
if t2_path.exists():
    with open(t2_path) as f:
        t2 = json.load(f)
    print("Table 2 -- Circuit Performance Metrics (reproduced)")
    print(f"{'Circuit':<20} {'Expressibility':>16} {'Entanglement':>14} {'Discreteness':>14}")
    print("-" * 70)
    for name, metrics in t2.items():
        marker = " <- SELECTED" if name == "Circuit-11" else ""
        print(f"{name:<20} {metrics.get('expressibility',0):>16.4f} "
              f"{metrics.get('entanglement',0):>14.4f} "
              f"{metrics.get('discreteness',0):>14.4f}{marker}")

# Table 3: Classification accuracy bar chart
t3_path = Path("results/experiment1/table3_circuit_classification.json")
if t3_path.exists():
    with open(t3_path) as f:
        t3 = json.load(f)
    circuits    = list(t3.keys())
    mnist_acc   = [t3[c].get("MNIST", 0) for c in circuits]
    fashion_acc = [t3[c].get("Fashion-MNIST", 0) for c in circuits]
    x = range(len(circuits))
    w = 0.35
    fig, ax = plt.subplots()
    ax.bar([i - w/2 for i in x], mnist_acc,   w, label="MNIST",         color='#4C78A8')
    ax.bar([i + w/2 for i in x], fashion_acc, w, label="Fashion-MNIST", color='#F58518')
    ax.set_xticks(list(x)); ax.set_xticklabels(circuits, rotation=45, ha='right', fontsize=9)
    ax.set_ylabel("Test Accuracy"); ax.set_ylim(0.4, 1.0)
    ax.set_title("Table 3 -- Circuit Classification Accuracy (Reproduced)")
    ax.legend()
    ax.axhline(y=0.8057, color='green', linestyle='--', linewidth=1)
    plt.tight_layout()
    plt.savefig("results/experiment1/table3_bar_chart.png", dpi=150)
    plt.show()
    print("Table 3 chart saved to results/experiment1/table3_bar_chart.png")
```

```python
# ============================================================
# Experiment 3 -- Display Results
# ============================================================
import json, numpy as np
from pathlib import Path
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['figure.figsize'] = (14, 5)

PAPER_REFS = {
    "data_noise":   {"0.0": 0.9005, "0.1": 0.8915, "0.2": 0.8900, "0.3": 0.8425},
    "bit_flip":     {"0.0": 0.9005, "0.1": 0.8769, "0.2": 0.8558, "0.3": 0.8405},
    "phase_flip":   {"0.0": 0.9005, "0.1": 0.8836, "0.2": 0.8618, "0.3": 0.8602},
    "depolarizing": {"0.0": 0.9005, "0.1": 0.8639, "0.2": 0.8664, "0.3": 0.8327},
}
x_labels     = ["0.0", "0.1", "0.2", "0.3"]
noise_labels = list(PAPER_REFS.keys())
table_labels = ["Table 5", "Table 6", "Table 7", "Table 8"]

noise_path = Path("results/experiment3/noise_results.json")
if noise_path.exists():
    with open(noise_path) as f:
        noise_res = json.load(f)

    fig, axes = plt.subplots(1, 4, figsize=(16, 4))
    fig.suptitle("Experiment 3 -- Noise Robustness (Reproduced vs Paper)", fontsize=13)
    for ax, noise_type, tbl in zip(axes, noise_labels, table_labels):
        ours  = noise_res.get(noise_type, {}).get("Proposed", {})
        paper = PAPER_REFS[noise_type]
        ours_v  = [float(ours.get(p, 0))  for p in x_labels]
        paper_v = [float(paper.get(p, 0)) for p in x_labels]
        xs = np.arange(len(x_labels))
        ax.bar(xs - 0.2, ours_v,  0.35, label="Reproduced", color='#4C78A8')
        ax.bar(xs + 0.2, paper_v, 0.35, label="Paper",      color='#F58518', alpha=0.7)
        ax.set_xticks(xs); ax.set_xticklabels([f"p={p}" for p in x_labels])
        ax.set_ylim(0.5, 1.0); ax.set_ylabel("Accuracy")
        ax.set_title(f"{noise_type.replace('_',' ').title()}\n({tbl})", fontsize=10)
        ax.legend(fontsize=8)
    plt.tight_layout()
    plt.savefig("results/experiment3/noise_comparison_all.png", dpi=150)
    plt.show()

    print("\nNoise Results vs Paper (Proposed Model)")
    print(f"{'Noise Type':<20} {'p':>5} {'Reproduced':>12} {'Paper':>10} {'Delta':>8}")
    print("-" * 60)
    for noise_type in noise_labels:
        ours  = noise_res.get(noise_type, {}).get("Proposed", {})
        paper = PAPER_REFS[noise_type]
        for p in x_labels:
            o = float(ours.get(p, 0))
            pp = float(paper.get(p, 0))
            print(f"{noise_type:<20} {p:>5} {o:>12.4f} {pp:>10.4f} {o-pp:>+8.4f}")
```

# Troubleshooting

### Token error: `Authentication failed`
- Go to https://github.com/settings/tokens?type=beta
- Make sure token has **Contents: Read and Write** permission on `QC-CNN-Parallel1`
- Paste the full token starting with `github_pat_` or `ghp_`

### `ModuleNotFoundError: No module named 'pennylane'`
- Re-run the Section 0.3 install cell

### `FileNotFoundError` for overhead_mnist
- Overhead-MNIST is not in torchvision. Download from: https://www.kaggle.com/datasets/andrewmvd/aircraft-mnist
- Place in `implementation/data/overhead_mnist/`
- Or remove `"overhead_mnist"` from `EXP2_DATASETS` in Section 0.2

### Experiment 3 very slow
- Set `EXP3_MAX_TEST_SAMPLES = 500` in Section 0.2 for a quick sanity check
- `default.mixed` has no GPU support -- this is a PennyLane limitation

### Session timeout (Colab)
- Enable a keep-alive browser extension
- Or switch to Kaggle (12 hr sessions) or AWS SageMaker Studio Lab (8 hr CPU)

### Experiment 1 too slow
- Set `RUN_CIRCUIT_METRICS = False` to skip the 5,000-simulation Table 2 analysis
- Set `N_SIMS = 500` for a quick test run
